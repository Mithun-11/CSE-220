## List

### Slicing 
a[start:end:step] end is not included

```python
a = [10, 20, 30, 40, 50]

a[1:4]   # [20, 30, 40]
a[:3]    # [10, 20, 30]
a[2:]    # [30, 40, 50]
a[::2]    # every 2nd item
a[::-1]   # reverse
a[1::2]   # indices 1, 3, 5...
```

### Add/remove

```python
a.append(50)        # add one item
a.extend([60, 70])  # add multiple items
a.insert(1, 15)     # insert at index 1

a.remove(20)   # remove value 20
a.pop()        # remove + return last item
a.pop(1)       # remove + return item at index 1
```

## Dictionary

```python
scores = {
    "Alice": 85,
    "Bob": 72,
    "Charlie": 91
}
```

### Access value
can be done using scores["Bob"] or scores.get("Bob") 
use the get one cause you can pass what to return if that is not present

```python
scores.get("Bob")
```

```python
scores.get("Ron",0)
#Here Ron isn't in the dictionary so we will get 0
```

### Add/Update

```python
scores["David"]=80
scores["Bob"]=70
```

### Remove

```python
scores.pop("Bob")
```

### Keys,values,items

```python
scores.keys() #Just the keys
scores.values() #Just the values
scores.items() # gets both key and value
```

```python
for name in scores.keys():
    print(name)
```

```python
for score in scores.values():
    print(score)
```

```python
for name,score in scores.items():
    print(name,score)
```

### Membership

```python
"Alice" in scores
```

```python
#This checks keys, not values.
85 in scores.values()
```

### Sorting Dictionaries

```python
#sorted by key
sorted(scores.items()) 
#sorted returns a list of tuples here
```

```python
# sort by value
sorted(scores.items(), key=lambda x: x[1])
```

```python
# sort by value in descending
sorted(scores.items(), key=lambda x: x[1],reverse=True)
```

Understand the lambda:

lambda x: x[1]

each x here is like

("Alice",85)

so x[0] is "Alice"

x[1] is 85

### Dictionary comprehension

```python
result = {
    name: score
    for name, score in scores.items()
    if score >= 85
}

print(result)
```

### Set Operations


```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
```

all the operations returns a new set

```python
A | B
# {1, 2, 3, 4, 5, 6}
A.union(B)
```

```python
A & B
# {3, 4}
A.intersection(B)
```

```python
A - B
# {1, 2}
```

```python
# Symmetric differnce(Item belongs in only one of the set)
A ^ B
# {1, 2, 5, 6}
```

## zip

```python
names = ["A", "B", "C"]
scores = [80, 90, 70]

#Conceptually zip(names,scores) pairs them like:
#("A", 80)
# ("B", 90)
# ("C", 70)

for name, score in zip(names, scores):
    print(name, score)

# it can be used with multiple list/anythin at once
# **a,b can be any iterable. They can also be differnt like list and tuple etc.
```
